```python
import matplotlib.pyplot as plt

class_counts = df["Class"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(
    ["Genuine (0)", "Fraud (1)"],
    class_counts.values
)

plt.title("Genuine vs Fraud Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

for i, value in enumerate(class_counts.values):
    plt.text(i, value, f"{value:,}", ha="center", va="bottom")

plt.show()
```

```python
import matplotlib.pyplot as plt

class_counts = df["Class"].value_counts().sort_index()

print("Transaction Distribution:")
print(class_counts)

print("\nTransaction Distribution (%):")
print((class_counts / len(df) * 100).round(3))

print("\nAmount Statistics by Transaction Type:")
print(
    df.groupby("Class")["Amount"]
      .agg(["count", "mean", "median", "min", "max"])
)

plt.figure(figsize=(8, 5))
plt.bar(
    ["Genuine (0)", "Fraud (1)"],
    class_counts.values
)

plt.title("Genuine vs Fraud Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

for i, value in enumerate(class_counts.values):
    plt.text(i, value, f"{value:,}", ha="center", va="bottom")

plt.show()

plt.figure(figsize=(10, 5))
plt.hist(df["Amount"], bins=50)

plt.title("Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()

fraud_amounts = df[df["Class"] == 1]["Amount"]

plt.figure(figsize=(10, 5))
plt.hist(fraud_amounts, bins=30)

plt.title("Fraud Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

df["Hour"] = (df["Time"] / 3600) % 24

amount_by_class = df.groupby("Class")["Amount"].agg(
    ["mean", "median", "min", "max"]
)

print("Amount Statistics by Class:")
display(amount_by_class)

fraud_by_hour = (
    df[df["Class"] == 1]
    .groupby("Hour")
    .size()
)

plt.figure(figsize=(10, 5))
plt.plot(fraud_by_hour.index, fraud_by_hour.values)

plt.title("Fraud Transactions by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Fraud Transactions")
plt.xticks(range(0, 24))

plt.show()

plt.figure(figsize=(10, 5))

plt.hist(
    df[df["Class"] == 0]["Amount"],
    bins=50,
    alpha=0.6,
    label="Genuine"
)

plt.hist(
    df[df["Class"] == 1]["Amount"],
    bins=50,
    alpha=0.6,
    label="Fraud"
)

plt.title("Transaction Amount: Genuine vs Fraud")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.legend()

plt.show()

fraud_df = df[df["Class"] == 1]

print("Fraud Transaction Summary:")
print("Number of fraud transactions:", len(fraud_df))
print("Average fraud amount:", round(fraud_df["Amount"].mean(), 2))
print("Median fraud amount:", round(fraud_df["Amount"].median(), 2))
print("Minimum fraud amount:", round(fraud_df["Amount"].min(), 2))
print("Maximum fraud amount:", round(fraud_df["Amount"].max(), 2))
```

```python
v_features = [f"V{i}" for i in range(1, 29)]

print("V1-V28 Statistical Summary:")
display(df[v_features].describe().T)

print("\nMean Feature Values by Class:")
feature_means = df.groupby("Class")[v_features].mean().T

display(feature_means)

feature_difference = (
    feature_means[1] - feature_means[0]
).sort_values(key=abs, ascending=False)

print("\nFeatures with Largest Mean Difference:")
display(feature_difference.head(10))

target_correlation = (
    df[v_features + ["Amount", "Class"]]
    .corr()["Class"]
    .drop("Class")
    .sort_values(key=abs, ascending=False)
)

print("\nTop Features Correlated with Class:")
display(target_correlation.head(10))
```

```python
# DATA CLEANING

import pandas as pd
import os

df_clean = pd.read_csv(
    "../data/raw/creditcard.csv/creditcard.csv"
)

print("Original dataset shape:", df_clean.shape)

missing_values = df_clean.isnull().sum().sum()

print("\nTotal missing values:", missing_values)

duplicate_count = df_clean.duplicated().sum()

print("Total duplicate rows:", duplicate_count)

df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("\nDataset shape after removing duplicates:", df_clean.shape)

print("\nData Types:")
print(df_clean.dtypes)

print("\nClass Distribution:")
print(df_clean["Class"].value_counts())

print("\nClass Distribution (%):")
print(
    (df_clean["Class"].value_counts(normalize=True) * 100).round(3)
)

print("\nAmount Statistics:")
print(df_clean["Amount"].describe())

print("\nTime Statistics:")
print(df_clean["Time"].describe())

print("\nFinal Validation:")
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())

processed_folder = "../data/processed"

os.makedirs(processed_folder, exist_ok=True)

output_path = os.path.join(
    processed_folder,
    "creditcard_clean.csv"
)

df_clean.to_csv(output_path, index=False)

print("\nClean dataset saved successfully!")
print("File:", output_path)
print("Final dataset shape:", df_clean.shape)
```

```python

# 1. Load cleaned dataset
clean_df = pd.read_csv("../data/processed/creditcard_clean.csv")

print("Cleaned Dataset Shape:", clean_df.shape)


# 2. Missing Values

print("\nMissing Values:")
print(clean_df.isnull().sum().sum())


# 3. Duplicate Check

print("\nDuplicate Rows:")
print(clean_df.duplicated().sum())

# 4. Class Distribution

print("\nClass Distribution:")
print(clean_df["Class"].value_counts())

print("\nClass Distribution (%):")
print(
    (clean_df["Class"].value_counts(normalize=True) * 100).round(3)
)

# 5. Fraud Transaction Count

fraud_count = (clean_df["Class"] == 1).sum()
genuine_count = (clean_df["Class"] == 0).sum()

print("\nGenuine Transactions:", genuine_count)
print("Fraud Transactions:", fraud_count)

# 6. Final Data Quality Check

print("\nFinal Data Quality Check:")

if clean_df.isnull().sum().sum() == 0:
    print("✓ No missing values")

if clean_df.duplicated().sum() == 0:
    print("✓ No duplicate rows")

if clean_df["Class"].isin([0, 1]).all():
    print("✓ Class values are valid")

if clean_df["Amount"].min() >= 0:
    print("✓ Transaction amounts are valid")

print("\nData cleaning validation completed successfully!")
```
