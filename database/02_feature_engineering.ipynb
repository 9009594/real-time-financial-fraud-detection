```python
import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_csv("../data/processed/creditcard_clean.csv")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nData Types:")
print(df.dtypes)
```

```python
df["Hour"] = (df["Time"] / 3600) % 24

def get_time_period(hour):
    if 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 18:
        return "Afternoon"
    elif 18 <= hour < 24:
        return "Evening"
    else:
        return "Night"

df["Time_Period"] = df["Hour"].apply(get_time_period)

print("New Time Features:")
display(df[["Time", "Hour", "Time_Period", "Class"]].head(10))

print("\nTime Period Distribution:")
print(df["Time_Period"].value_counts())

print("\nFraud by Time Period:")
print(
    pd.crosstab(
        df["Time_Period"],
        df["Class"],
        normalize="index"
    ).round(4) * 100
)
```

```python
print("Original Amount Statistics:")
display(df["Amount"].describe())

print("\nOriginal Amount Skewness:")
print(round(df["Amount"].skew(), 4))

df["Log_Amount"] = np.log1p(df["Amount"])

print("\nAmount Feature Comparison:")
display(
    df[["Amount", "Log_Amount", "Class"]].describe()
)

print("\nLog Amount Skewness:")
print(round(df["Log_Amount"].skew(), 4))

print("\nSample Amount Transformation:")
display(
    df[["Amount", "Log_Amount", "Class"]].head(10)
)
```

```python
print("Dataset Shape After Feature Engineering:")
print(df.shape)

print("\nAll Columns:")
print(df.columns.tolist())

new_features = ["Hour", "Time_Period", "Log_Amount"]

print("\nNew Features:")
display(df[new_features].head())

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Time Periods:")
print(df["Time_Period"].unique())

print("\nHour Range:")
print(df["Hour"].min(), "to", df["Hour"].max())

numeric_features = df.select_dtypes(include=np.number)

correlation_with_class = (
    numeric_features.corr()["Class"]
    .drop("Class")
    .sort_values(key=abs, ascending=False)
)

print("\nTop Numeric Features Correlated with Class:")
display(correlation_with_class.head(15))

print("\nFinal Feature Engineering Validation:")

if df.isnull().sum().sum() == 0:
    print("✓ No missing values")

if df.duplicated().sum() == 0:
    print("✓ No duplicate rows")

if df["Hour"].between(0, 24, inclusive="left").all():
    print("✓ Hour feature is valid")

if (df["Log_Amount"] >= 0).all():
    print("✓ Log_Amount feature is valid")

print("\nFeature engineering validation completed!")
```

```python
output_path = "../data/processed/creditcard_feature_engineered.csv"

df.to_csv(output_path, index=False)

print("Feature-engineered dataset saved successfully!")
print("File:", output_path)
print("Final Shape:", df.shape)

# Verify saved file
saved_df = pd.read_csv(output_path)

print("\nVerification:")
print("Saved Shape:", saved_df.shape)
print("Missing Values:", saved_df.isnull().sum().sum())
print("Duplicate Rows:", saved_df.duplicated().sum())
```
