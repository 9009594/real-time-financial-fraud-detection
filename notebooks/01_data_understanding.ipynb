```python
import pandas as pd

DATA_PATH = "../data/raw/creditcard.csv/creditcard.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Total transactions:", len(df))
```

```python
# Check fraud and normal transactions

fraud_counts = df["Class"].value_counts()

print("Normal transactions:", fraud_counts[0])
print("Fraud transactions:", fraud_counts[1])
```

```python
# Calculate fraud percentage

total_transactions = len(df)
fraud_transactions = df["Class"].sum()

fraud_percentage = (fraud_transactions / total_transactions) * 100

print("Total transactions:", total_transactions)
print("Fraud transactions:", fraud_transactions)
print("Fraud percentage:", round(fraud_percentage, 4), "%")
```

```python
# Check dataset shape

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Check missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# Check duplicate rows

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
duplicate_percentage = (duplicate_count / len(df)) * 100

print("Duplicate rows:", duplicate_count)
print("Duplicate percentage:", round(duplicate_percentage, 4), "%")
```

```python
import matplotlib.pyplot as plt

# Count normal and fraud transactions
class_counts = df["Class"].value_counts()

plt.figure(figsize=(6, 4))
class_counts.plot(kind="bar")

plt.title("Normal vs Fraud Transactions")
plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")
plt.xticks([0, 1], ["Normal", "Fraud"], rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Amount"], bins=50)

plt.title("Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        df[df["Class"] == 0]["Amount"],
        df[df["Class"] == 1]["Amount"]
    ],
    tick_labels=["Normal", "Fraud"]
)

plt.title("Transaction Amount: Normal vs Fraud")
plt.xlabel("Transaction Type")
plt.ylabel("Amount")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["Time"], bins=50)

plt.title("Transaction Time Distribution")
plt.xlabel("Time")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
fraud_df = df[df["Class"] == 1]

plt.figure(figsize=(10, 5))

plt.scatter(
    fraud_df["Time"],
    fraud_df["Amount"],
    alpha=0.6
)

plt.title("Fraud Transactions Over Time")
plt.xlabel("Time (seconds)")
plt.ylabel("Transaction Amount")

plt.show()
```
