```python
import pandas as pd

DATA_PATH = "../data/raw/creditcard.csv/creditcard.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Correlation with fraud class

correlation = df.corr(numeric_only=True)["Class"].sort_values(ascending=False)

print("Correlation with Class:")
print(correlation)
```

```python
import matplotlib.pyplot as plt

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(14, 10))

plt.imshow(correlation_matrix, cmap="coolwarm", aspect="auto")
plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Feature Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
# Create Hour feature from Time
df["Hour"] = (df["Time"] / 3600) % 24

# Create Time Period feature
def get_time_period(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["Time_Period"] = df["Hour"].apply(get_time_period)

# Create Log Amount feature
import numpy as np

df["Log_Amount"] = np.log1p(df["Amount"])

df[["Time", "Hour", "Time_Period", "Amount", "Log_Amount", "Class"]].head()
```

```python
df[["Hour", "Amount", "Log_Amount"]].describe()
df["Time_Period"].value_counts()
df.groupby("Time_Period")["Class"].agg(["count", "sum"])
```

```python
time_fraud_rate = (
    df.groupby("Time_Period")["Class"]
    .agg(["count", "sum"])
    .rename(columns={"count": "Transactions", "sum": "Fraud"})
)

time_fraud_rate["Fraud_Rate_%"] = (
    time_fraud_rate["Fraud"] / time_fraud_rate["Transactions"] * 100
)

time_fraud_rate.sort_values("Fraud_Rate_%", ascending=False)
```

```python
import matplotlib.pyplot as plt

time_fraud_rate["Fraud_Rate_%"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Fraud Rate by Time Period")
plt.xlabel("Time Period")
plt.ylabel("Fraud Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
