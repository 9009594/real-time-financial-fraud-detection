```python
import pandas as pd

DATA_PATH = "../data/processed/creditcard_clean.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
print("Current columns:")
print(df.columns.tolist())

print("\nDataset Shape:")
print(df.shape)
```

```python
df["Hour"] = (df["Time"] / 3600) % 24

df["Hour"] = df["Hour"].astype(int)

print("Hour feature created successfully!")
print(df[["Time", "Hour"]].head())
```

```python
def get_time_period(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"


df["Time_Period"] = df["Hour"].apply(get_time_period)

print("Time_Period feature created successfully!")
print(df["Time_Period"].value_counts())
```

```python
import numpy as np

df["Log_Amount"] = np.log1p(df["Amount"])

print("Log_Amount feature created successfully!")
print(df[["Amount", "Log_Amount"]].head())
```
